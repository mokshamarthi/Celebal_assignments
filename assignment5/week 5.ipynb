```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.types import *

spark = SparkSession.builder.appName("Week5").getOrCreate()
```

```python
df = spark.read.csv(
    "/Volumes/week5data/default/week5data/week5_dataset.csv",
    header=True,
    inferSchema=True
)
```

```python
display(df)
```

```python
df_q3 = df.dropDuplicates(["user_id", "transaction_date"])
display(df_q3)
```

```python
df_q4 = df.filter(col("region") == "West") \
    .groupBy("product_category") \
    .agg(avg("sale_amount").alias("average_sale"))
display(df_q4)
```

```python
df_q5 = df.na.fill({"status": "Unknown"})
display(df_q5)
```

```python
df_q6 = df.groupBy("city") \
          .count() \
          .filter(col("count") > 100)

display(df_q6)
```

```python
df_new = df.drop("quantity")
df_new = df_new.withColumnRenamed("price", "product_price")
```

```python
df_q8 = df.filter(
    (col("age").between(18, 30)) &
    (col("subscription") == "Premium")
)
display(df_q8)
```

```python
df_q10 = df.withColumn(
    "event_time",
    col("raw_timestamp").cast(TimestampType())
).drop("raw_timestamp")
display(df_q10)
```

```python
df_q12 = df.filter(
    col("email").isNotNull() &
    (trim(col("username")) != "")
)
display(df_q12)
```

```python
df_q13 = df.agg(
    min("price").alias("Minimum_Price"),
    max("price").alias("Maximum_Price"),
    avg("price").alias("Average_Price")
)
display(df_q13)
```

```python
from pyspark.sql.functions import sum

df_q15 = (
    df.dropDuplicates()
      .na.fill({"price": 0})
      .groupBy("store_id")
      .agg(sum("price").alias("total_revenue"))
)
display(df_q15)
```
