```python
# 3.Handle missing values (identify, fill/drop). 
```

```python
df.isnull().sum()
```

```python
df["discount"]= df["discount"].fillna(df["discount"].mean())
```

```python
df["seller_information"]=(df.groupby("seller_name")["seller_information"].transform(lambda x:x.ffill().bfill()))
```

```python
cols=["what_customers_said","seller_name","videos","seller_information","variations"]
df[cols]=df[cols].fillna("Not Available")
```

```python
df.drop_duplicates(inplace=True)
```

```python
#6.Create a derived column (total_amount = price * quantity).
```

```python
df.insert(9,"quantity",(df["rating"]*3 + df["discount"]/10).clip(lower=1).astype(int))
```

```python
df.head(3)
```

```python
df["final_price"].dtype
```

```python
df["final_price"]=pd.to_numeric(df["final_price"].replace(r'[₹,"]','',regex=True), errors="coerce")
```

```python
df["final_price"].dtype
```

```python
df.insert(10,"total_amount",(df["final_price"]*df["quantity"]))
df.head(4)
```
