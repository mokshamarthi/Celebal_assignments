```python
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/week6/default/week6/week6_spark_dataset.csv")
```

```python
display(df)
```

```python
electronics_df = df.filter(df.category == "Electronics") \
                   .select("product_id", "price")
display(electronics_df)
```

```python
from pyspark.sql.functions import col
revised_df = df.withColumnRenamed("old_name", "new_name") \
               .withColumn("price", col("price").cast("double"))
display(revised_df)
```

```python
completed_orders = df.filter(
    (df.status == "Completed") &
    (df.amount > 1000)
)
display(completed_orders)
```

```python
from pyspark.sql.functions import col
final_df = df.withColumn(
    "final_price",
    col("base_price") * 1.18
)
display(final_df)
```

```python
from pyspark.sql.functions import col
clean_df = df.filter(col("user_id").isNotNull())
clean_df.write \
    .mode("overwrite") \
    .parquet("/Volumes/week6/default/week6/output_parquet")
```

```python
parquet_df = spark.read.parquet("/Volumes/week6/default/week6/output_parquet")
display(parquet_df)
```

```python
parquet_df.write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/Volumes/week6/default/week6/output_csv")
display(parquet_df)
```

```python
north_high = df.filter(
    (df.region == "North") |
    (df.priority == "High")
)
display(north_high)
```
